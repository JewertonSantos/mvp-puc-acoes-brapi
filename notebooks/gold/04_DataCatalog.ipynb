```python
%sql
-- UNION ALL reúne os domínios observados dos diferentes critérios.
-- A coluna "criterio" identifica de qual campo veio cada valor.

SELECT 'criterio_dy' AS criterio, criterio_dy AS valor
FROM workspace.gold.criterios_acoes

UNION ALL

SELECT 'criterio_crescimento', criterio_crescimento
FROM workspace.gold.criterios_acoes

UNION ALL

SELECT 'criterio_pl_crescimento', criterio_pl_crescimento
FROM workspace.gold.criterios_acoes

UNION ALL

SELECT 'criterio_pvp_x_pl', criterio_pvp_x_pl
FROM workspace.gold.criterios_acoes

UNION ALL

SELECT 'criterio_margem', criterio_margem
FROM workspace.gold.criterios_acoes

UNION ALL

SELECT 'criterio_roe', criterio_roe
FROM workspace.gold.criterios_acoes

UNION ALL

SELECT 'criterio_divida', criterio_divida
FROM workspace.gold.criterios_acoes

UNION ALL

SELECT 'criterio_bazin', criterio_bazin
FROM workspace.gold.criterios_acoes

UNION ALL

SELECT 'criterio_graham', criterio_graham
FROM workspace.gold.criterios_acoes

ORDER BY criterio, valor;
```
