```python
%sql
CREATE SCHEMA IF NOT EXISTS workspace.silver;
```

```python
from pyspark.sql import functions as F

bronze = spark.table("workspace.bronze.cotacoes")

silver = (
    bronze
    .withColumn("symbol", F.upper(F.trim(F.col("symbol"))))
    .withColumn("date", F.to_date(F.col("date")))
    .withColumn("close", F.col("close").cast("double"))
    .withColumn("adjusted_close", F.col("adjusted_close").cast("double"))
    .filter(
        F.col("symbol").isNotNull() &
        F.col("date").isNotNull()
    )
    .select(
        "symbol",
        "date",
        "close",
        "adjusted_close"
    )
    .dropDuplicates(["symbol", "date"])
)

silver.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("workspace.silver.cotacoes")

display(spark.table("workspace.silver.cotacoes"))

spark.table("workspace.silver.cotacoes").printSchema()
```

```python
%sql
COMMENT ON TABLE workspace.silver.cotacoes
IS 'Cotações históricas diárias padronizadas dos ativos entre 2020 e 2026.';

ALTER TABLE workspace.silver.cotacoes
ALTER COLUMN symbol
COMMENT 'Código ou ticker padronizado do ativo.';

ALTER TABLE workspace.silver.cotacoes
ALTER COLUMN date
COMMENT 'Data da cotação no formato DATE.';

ALTER TABLE workspace.silver.cotacoes
ALTER COLUMN close
COMMENT 'Preço de fechamento do ativo no dia.';

ALTER TABLE workspace.silver.cotacoes
ALTER COLUMN adjusted_close
COMMENT 'Preço de fechamento ajustado.';
```

```python
from pyspark.sql import functions as F

bronze = spark.table("workspace.bronze.estatisticas")

silver = (
    bronze
    .withColumn("symbol", F.upper(F.trim(F.col("symbol"))))
    .withColumn("date", F.to_date(F.col("date")))
    .withColumn("price", F.col("price").cast("double"))
    .withColumn("bookValue", F.col("bookValue").cast("double"))
    .withColumn("priceToBook", F.col("priceToBook").cast("double"))
    .withColumn("trailingEps", F.col("trailingEps").cast("double"))
    .withColumn("trailingPE", F.col("trailingPE").cast("double"))
    .withColumn("dividendYield", F.col("dividendYield").cast("double"))
    .filter(
        F.col("symbol").isNotNull() &
        F.col("date").isNotNull()
    )
    .select(
        "symbol",
        "date",
        "period",
        "type",
        "price",
        "bookValue",
        "priceToBook",
        "trailingEps",
        "trailingPE",
        "dividendYield"
    )
    .dropDuplicates(["symbol", "date", "period", "type"])
)

silver.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("workspace.silver.estatisticas")

display(spark.table("workspace.silver.estatisticas"))

spark.table("workspace.silver.estatisticas").printSchema()
```

```python
from pyspark.sql import functions as F

bronze = spark.table("workspace.bronze.dados_financeiros")

silver = (
    bronze
    .withColumn("symbol", F.upper(F.trim(F.col("symbol"))))
    .withColumn("date", F.to_date(F.col("date")))
    .withColumn("totalCash", F.col("totalCash").cast("double"))
    .withColumn("ebitda", F.col("ebitda").cast("double"))
    .withColumn("totalDebt", F.col("totalDebt").cast("double"))
    .withColumn("returnOnEquity", F.col("returnOnEquity").cast("double"))
    .withColumn("profitMargins", F.col("profitMargins").cast("double"))
    .filter(
        F.col("symbol").isNotNull() &
        F.col("date").isNotNull()
    )
    .select(
        "symbol",
        "date",
        "period",
        "type",
        "totalCash",
        "ebitda",
        "totalDebt",
        "returnOnEquity",
        "profitMargins"
    )
    .dropDuplicates(["symbol", "date", "period", "type"])
)

silver.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("workspace.silver.dados_financeiros")

display(spark.table("workspace.silver.dados_financeiros"))
spark.table("workspace.silver.dados_financeiros").printSchema()
```

```python
from pyspark.sql import functions as F

bronze = spark.table("workspace.bronze.dividendos")

silver = (
    bronze
    .withColumn("symbol", F.upper(F.trim(F.col("symbol"))))
    .withColumn("eventType", F.trim(F.col("eventType")))
    .withColumn("label", F.trim(F.col("label")))
    .withColumn("rate", F.col("rate").cast("double"))
    .withColumn("approvedOn", F.to_date(F.col("approvedOn")))
    .withColumn("lastDatePrior", F.to_date(F.col("lastDatePrior")))
    .withColumn("exDate", F.to_date(F.col("exDate")))
    .withColumn("paymentDate", F.to_date(F.col("paymentDate")))
    .withColumn("relatedTo", F.trim(F.col("relatedTo")))
    .filter(
        F.col("symbol").isNotNull() &
        F.col("rate").isNotNull()
    )
    .select(
        "symbol",
        "eventType",
        "label",
        "rate",
        "approvedOn",
        "lastDatePrior",
        "exDate",
        "paymentDate",
        "relatedTo"
    )
    .dropDuplicates()
)

silver.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("workspace.silver.dividendos")
display(spark.table("workspace.silver.dividendos"))
spark.table("workspace.silver.dividendos").printSchema()
```

```python
%sql
COMMENT ON TABLE workspace.silver.dividendos
IS 'Eventos de dividendos e juros sobre capital próprio padronizados para análise histórica de proventos dos ativos.';

ALTER TABLE workspace.silver.dividendos
ALTER COLUMN symbol
COMMENT 'Código ou ticker padronizado do ativo.';

ALTER TABLE workspace.silver.dividendos
ALTER COLUMN eventType
COMMENT 'Tipo do evento de provento informado pela fonte.';

ALTER TABLE workspace.silver.dividendos
ALTER COLUMN label
COMMENT 'Descrição ou identificação do evento de provento.';

ALTER TABLE workspace.silver.dividendos
ALTER COLUMN rate
COMMENT 'Valor do provento por ação informado pela fonte.';

ALTER TABLE workspace.silver.dividendos
ALTER COLUMN approvedOn
COMMENT 'Data de aprovação do provento.';

ALTER TABLE workspace.silver.dividendos
ALTER COLUMN lastDatePrior
COMMENT 'Última data anterior utilizada como referência para o evento.';

ALTER TABLE workspace.silver.dividendos
ALTER COLUMN exDate
COMMENT 'Data ex-provento do evento.';

ALTER TABLE workspace.silver.dividendos
ALTER COLUMN paymentDate
COMMENT 'Data prevista ou realizada de pagamento do provento.';

ALTER TABLE workspace.silver.dividendos
ALTER COLUMN relatedTo
COMMENT 'Período ou referência ao qual o provento está relacionado.';
```
